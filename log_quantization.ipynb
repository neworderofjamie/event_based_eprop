```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot(num_levels, max_x):
    x = np.arange(max_x)

    log_scale = max_x / np.log(max_x + 1.0)
    
    y_levels = np.linspace(0.0, max_x, num_levels)
    x_levels = np.exp(y_levels / log_scale) - 1.0
    
    fig,axis = plt.subplots()
    axis.plot(x, log_scale * np.log(x + 1.0))
    axis.plot(x,x)
    
    axis.hlines(y_levels, 0, x_levels, linestyle="--", color="gray")
    axis.vlines(x_levels, 0, y_levels, linestyle="--", color="gray")

    
    print(x_levels)
```

```python
plot(3, 30.0)
```

```python
plot(9, 30.0)
```

```python
plot(9, 100.0)
```

```python
def calc(x, num_levels):
    if x == 0.0:
        return x
    else:
        scale = np.exp(num_levels - 1.0)
        y = (1.0 / scale) * np.exp(np.round(np.log(np.round(abs(x) * scale))))
        return np.copysign(y, x)
```

```python
x_range = np.linspace(-1.0, 1.0, 200)
fig,axis = plt.subplots(figsize=(16,8))
for l in range(1, 6):
    
    y = [calc(x, l) for x in x_range]
    axis.plot(x_range, y, label=f"({l} levels) - {len(np.unique(y))} unique values")
axis.plot(x_range, x_range, linestyle="--")
axis.legend()
fig.tight_layout(pad=0)
```
