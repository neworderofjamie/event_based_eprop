```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot(num_levels, max_x):
    x = np.arange(max_x)

    log_scale = max_x / np.log(max_x + 1.0)
    
    y_levels = np.linspace(0.0, max_x, num_levels)
    x_levels = np.exp(y_levels / log_scale) - 1.0
    
    fig,axis = plt.subplots()
    axis.plot(x, log_scale * np.log(x + 1.0))
    axis.plot(x,x)
    
    axis.hlines(y_levels, 0, x_levels, linestyle="--", color="gray")
    axis.vlines(x_levels, 0, y_levels, linestyle="--", color="gray")

    print(x_levels)
```

```python
plot(3, 30.0)
```

```python
plot(5, 30.0)
```

```python
plot(5, 100.0)
```
